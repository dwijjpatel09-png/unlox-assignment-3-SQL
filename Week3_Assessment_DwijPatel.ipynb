```python
import subprocess

def run_sql(query):
    result = subprocess.run(
        ["mysql", "-D", "flipcart", "-e", query],
        capture_output=True,
        text=True
    )

    print(result.stdout)

    if result.stderr:
        print("ERROR:")
        print(result.stderr)
```

```python
run_sql("""
SELECT *
FROM products;
""")
```

```python
run_sql("""
SELECT *
FROM products;
""")
```

```python
run_sql("""
SELECT product_name, price
FROM products
WHERE category = '_____';
""")
```

```python
run_sql("""
SELECT product_name, price
FROM products
WHERE category = 'Books';
""")
```

```python
run_sql("""
SELECT *
FROM products
WHERE price > 10000
ORDER BY price DESC;
""")
```

```python
run_sql("""
SELECT product_name, price
FROM products
WHERE category = 'Electronics'
ORDER BY price DESC
LIMIT 5;
""")
```

```python
run_sql("""
SELECT DISTINCT category
FROM products;
""")
```

```python
run_sql("""
SELECT COUNT(*) AS total_products
FROM products;
""")
```

```python
run_sql("""
SELECT AVG(price) AS average_price
FROM products
WHERE category = 'Books';
""")
```

```python
run_sql("""
SELECT
    MAX(price) AS maximum_price,
    MIN(price) AS minimum_price
FROM products;
""")
```

```python
run_sql("""
SELECT
    category,
    COUNT(*) AS product_count
FROM products
GROUP BY category;
""")
```

```python
run_sql("""
SELECT
    category,
    SUM(stock_quantity) AS total_stock
FROM products
GROUP BY category;
""")
```

```python
run_sql("""
SELECT
    category,
    AVG(price) AS average_price
FROM products
GROUP BY category
ORDER BY average_price DESC;
""")
```

```python
run_sql("""
SELECT
    brand,
    COUNT(*) AS product_count,
    AVG(price) AS average_price
FROM products
GROUP BY brand
HAVING COUNT(*) > 1;
""")
```

```python
run_sql("""
SELECT
    category,
    COUNT(*) AS active_product_count
FROM products
WHERE is_active = TRUE
GROUP BY category
HAVING COUNT(*) > 4;
""")
```

```python
run_sql("""
SELECT *
FROM products
ORDER BY price DESC
LIMIT 3;
""")
```

```python
run_sql("""
SELECT
    category,
    AVG(price) AS average_price
FROM products
GROUP BY category
HAVING AVG(price) > 2000;
""")
```

```python
run_sql("""
SELECT *
FROM products
WHERE avg_rating IS NULL;
""")
```

```python
run_sql("""
SELECT
    product_name,
    COALESCE(avg_rating, 'New Launch') AS rating
FROM products;
""")
```

```python
run_sql("""
SELECT
    product_name,
    price,
    CASE
        WHEN price < 1000 THEN 'Budget'
        WHEN price < 10000 THEN 'Mid'
        ELSE 'Premium'
    END AS price_tier
FROM products;
""")
```

```python
run_sql("""
SELECT
    category,
    COUNT(*) AS total_products,
    SUM(
        CASE
            WHEN price >= 10000 THEN 1
            ELSE 0
        END
    ) AS premium_products
FROM products
GROUP BY category;
""")
```

```python
run_sql("""
SELECT
    category,
    COUNT(*) AS total_products,

    SUM(
        CASE
            WHEN is_active = TRUE THEN 1
            ELSE 0
        END
    ) AS active_products,

    AVG(price) AS average_price,

    CASE
        WHEN AVG(price) < 1000 THEN 'Budget'
        WHEN AVG(price) < 10000 THEN 'Mid'
        ELSE 'Premium'
    END AS category_tier

FROM products

GROUP BY category

HAVING COUNT(*) >= 3

ORDER BY average_price DESC;
""")
```
